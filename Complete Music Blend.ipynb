### Predicting Popularity

```python
# LINEAR REGRESSION

features = [
    'danceability','energy','valence','tempo',
    'acousticness','instrumentalness',
    'liveness','speechiness','loudness'
]

X = dtf2[features]
y = dtf2['popularity']
```

```python
dtf2.isna().sum()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.4, random_state=42
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X_train_scaled, y_train)

y_pred_lr = lr.predict(X_test_scaled)
```

```python
plt.scatter(X_test['energy'], y_test, c="g")
plt.scatter(X_test['energy'], y_pred_lr, c="r")
plt.xlabel("Energy")
plt.ylabel("Popularity")
plt.show()
```

```python
def evaluate(y_true, y_pred, name):
    from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
    import numpy as np

    print(f"📊 Model: {name}")
    
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)

    print(f"MAE  : {mae:.2f}")
    print(f"MSE  : {mse:.2f}")
    print(f"RMSE : {rmse:.2f}")
    print(f"R²   : {r2:.4f}")
    print("-" * 30)
```

```python
evaluate(y_test, y_pred_lr, "Linear Regression")
```

```python
lr.coef_
```

```python
from sklearn.linear_model import ElasticNet
from sklearn.model_selection import GridSearchCV

params = {
    'alpha': [0.1, 1, 10],
    'l1_ratio': [0.2, 0.5, 0.8]
}

enet = ElasticNet()

grid = GridSearchCV(enet, params, cv=5)
grid.fit(X_train_scaled, y_train)

best_model = grid.best_estimator_
y_pred_enet = best_model.predict(X_test_scaled)
```

```python
evaluate(y_test, y_pred_enet, "ElasticNet")
```

```python
def get_cluster_overlap(input_features, scaler, kmeans, audio_feature_count):
    """
    input_features: list of audio features (length = number of audio features)
    scaler: fitted StandardScaler
    kmeans: trained KMeans model
    audio_feature_count: number of audio features used during training
    """

    # Step 1: Convert to array
    x = np.array(input_features).reshape(1, -1)

    # Step 2: Pad with zeros for genre columns
    total_features = kmeans.cluster_centers_.shape[1]
    padded = np.zeros((1, total_features))
    padded[0, :audio_feature_count] = x

    # Step 3: Scale
    x_scaled = scaler.transform(padded)

    # Step 4: Get cluster centers
    centers = kmeans.cluster_centers_

    # Step 5: Compute distances
    distances = np.linalg.norm(centers - x_scaled, axis=1)

    # Step 6: Convert to similarity (inverse distance)
    similarities = 1 / (distances + 1e-8)

    # Step 7: Normalize to percentage
    percentages = similarities / similarities.sum() * 100

    # Step 8: Return result
    return {f"Cluster {i}": round(p, 2) for i, p in enumerate(percentages)}
```

```python
def get_knn_cluster_overlap(input_features, scaler, knn_model, dtf, audio_feature_count, k=20):
    """
    input_features: list of audio features
    scaler: fitted scaler
    knn_model: fitted NearestNeighbors model
    dtf: dataframe containing 'vibe_cluster'
    audio_feature_count: number of audio features
    k: number of neighbors
    """

    # Step 1: prepare input
    x = np.array(input_features).reshape(1, -1)

    # Step 2: pad zeros for genre columns
    total_features = knn_model._fit_X.shape[1]
    padded = np.zeros((1, total_features))
    padded[0, :audio_feature_count] = x

    # Step 3: scale
    x_scaled = scaler.transform(padded)

    # Step 4: find neighbors
    distances, indices = knn_model.kneighbors(x_scaled, n_neighbors=k)

    # Step 5: get cluster labels of neighbors
    neighbor_clusters = dtf.iloc[indices[0]]['vibe_cluster']

    # Step 6: compute percentages
    cluster_counts = neighbor_clusters.value_counts(normalize=True) * 100

    # Step 7: return result
    return {f"Cluster {int(c)}": round(p, 2) for c, p in cluster_counts.items()}
```

```python
def get_vibe_probabilities(audio_input, scaler, model):
    """
    Returns the probability of a song belonging to each of the 6 vibes.
    """
    # 1. Prepare and Scale Input
    features = ['danceability', 'energy', 'loudness', 'speechiness', 'acousticness', 
                'instrumentalness', 'liveness', 'valence', 'tempo']
    input_array = np.array([[audio_input[f] for f in features]])
    scaled_input = scaler.transform(input_array)
    
    # 2. Get Probabilities from Logistic Regression
    # predict_proba returns an array of 6 percentages
    probs = model.predict_proba(scaled_input)[0]
    
    # 3. Format Output
    return {f"Cluster {i}": f"{round(p*100, 2)}%" for i, p in enumerate(probs)}

# Example: Check a "Pop-Rock" hybrid song
sample_hybrid = {
    'danceability': 0.65, 'energy': 0.75, 'loudness': -6.0, 
    'speechiness': 0.04, 'acousticness': 0.15, 'instrumentalness': 0.1, 
    'liveness': 0.1, 'valence': 0.5, 'tempo': 120.0
}

print(get_vibe_probabilities(sample_hybrid, scaler, log_reg))
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def calculate_user_similarity(user1_df, user2_df, scaler, log_reg_model):
    # 1. Select the 9 audio features
    features = ['danceability', 'energy', 'loudness', 'speechiness', 'acousticness', 
                'instrumentalness', 'liveness', 'valence', 'tempo']
    
    # 2. Get Soft Cluster Probabilities for all songs in User 1's history
    u1_scaled = scaler.transform(user1_df[features].fillna(0))
    u1_probs = log_reg_model.predict_proba(u1_scaled)
    # Average the probabilities to get the "Global Taste Vector"
    u1_profile = u1_probs.mean(axis=0).reshape(1, -1)
    
    # 3. Repeat for User 2
    u2_scaled = scaler.transform(user2_df[features].fillna(0))
    u2_probs = log_reg_model.predict_proba(u2_scaled)
    u2_profile = u2_probs.mean(axis=0).reshape(1, -1)
    
    # 4. Calculate Cosine Similarity between the two profiles
    similarity_score = cosine_similarity(u1_profile, u2_profile)[0][0]
    
    return similarity_score, u1_profile, u2_profile

# --- Execute Analysis ---
score, prof1, prof2 = calculate_user_similarity(user1, user2, scaler, log_reg)

print(f"Total Music Taste Match: {score * 100:.2f}%")
```

#### RANDOM FOREST

```python
def compute_rf_user_similarity(user1_df, user2_df, scaler, rf_model):
    """
    Computes the % match between two users using Random Forest probabilities.
    """
    # 1. Define the 9 audio features used by the model
    features = ['danceability', 'energy', 'loudness', 'speechiness', 'acousticness', 
                'instrumentalness', 'liveness', 'valence', 'tempo']
    
    # 2. Generate the "Taste Fingerprint" for User 1
    # Scale the features first
    u1_scaled = scaler.transform(user1_df[features].fillna(0))
    # Get probabilities for every song (e.g., 50 songs x 6 clusters)
    u1_song_probs = rf_model.predict_proba(u1_scaled)
    # Average them to get the 6-dimensional User Profile
    u1_profile = u1_song_probs.mean(axis=0).reshape(1, -1)
    
    # 3. Generate the "Taste Fingerprint" for User 2
    u2_scaled = scaler.transform(user2_df[features].fillna(0))
    u2_song_probs = rf_model.predict_proba(u2_scaled)
    u2_profile = u2_song_probs.mean(axis=0).reshape(1, -1)
    
    # 4. Calculate Cosine Similarity between the two 6D vectors
    # This formula calculates: (A . B) / (||A|| ||B||)
    similarity_score = cosine_similarity(u1_profile, u2_profile)[0][0]
    
    return similarity_score, u1_profile, u2_profile

# --- Execution ---
# Assuming 'rf_classifier' and 'scaler' are already trained from your previous step
match_score, p1, p2 = compute_rf_user_similarity(user1, user2, scaler, rf_classifier)

print(f"Random Forest Taste Match Score: {match_score * 100:.2f}%")
```

#### XGBOOST

```python
def compute_xgboost_user_similarity(user1_df, user2_df, scaler, xgb_model):
    """
    Calculates the similarity score between two users based on XGBoost cluster probabilities.
    """
    # 1. Define the audio features used during training
    features = ['danceability', 'energy', 'loudness', 'speechiness', 'acousticness', 
                'instrumentalness', 'liveness', 'valence', 'tempo']
    
    # 2. Extract and Scale features for both users
    u1_scaled = scaler.transform(user1_df[features].fillna(0))
    u2_scaled = scaler.transform(user2_df[features].fillna(0))
    
    # 3. Get 'Soft' Cluster Assignments (Probabilities)
    # This returns an array of shape (n_songs, 6_clusters)
    u1_probs = xgb_model.predict_proba(u1_scaled)
    u2_probs = xgb_model.predict_proba(u2_scaled)
    
    # 4. Create the Mean Taste Vector (The User Fingerprint)
    # We average the probabilities across all songs in their history
    u1_fingerprint = np.mean(u1_probs, axis=0).reshape(1, -1)
    u2_fingerprint = np.mean(u2_probs, axis=0).reshape(1, -1)
    
    # 5. Compute Cosine Similarity
    # Range is 0 (completely different) to 1 (identical distribution)
    similarity_score = cosine_similarity(u1_fingerprint, u2_fingerprint)[0][0]
    
    return similarity_score, u1_fingerprint, u2_fingerprint

# --- Usage ---
match_percent, profile1, profile2 = compute_xgboost_user_similarity(user1, user2, scaler, xgb_vibe_model)
print(f"User Match Score: {match_percent * 100:.2f}%")
```

# Recommendation System

```python
def recommend_songs(user_df, main_pool_df, scaler, vibe_model, top_n=10):
    # 1. Identify User's Favorite Vibe
    # We use the model to find which cluster the user listens to most
    features = ['danceability', 'energy', 'loudness', 'speechiness', 'acousticness', 
                'instrumentalness', 'liveness', 'valence', 'tempo']
    
    u_scaled = scaler.transform(user_df[features].fillna(0))
    u_clusters = vibe_model.predict(u_scaled)
    fav_cluster = pd.Series(u_clusters).mode()[0] # The most frequent cluster
    
    # 2. Filter Main Pool for that Vibe
    # Only look at songs that haven't been heard and match the favorite cluster
    candidate_pool = main_pool_df[
        (main_pool_df['vibe_cluster'] == fav_cluster) & 
        (~main_pool_df['track_name'].isin(user_df['track_name']))
    ]
    
    # 3. Calculate "Fine-Tune" Similarity
    # Find songs in that cluster that specifically match the user's average audio taste
    u_avg_vector = u_scaled.mean(axis=0).reshape(1, -1)
    pool_scaled = scaler.transform(candidate_pool[features].fillna(0))
    
    candidate_pool['match_score'] = cosine_similarity(pool_scaled, u_avg_vector).flatten()
    
    # 4. Return Top N
    return candidate_pool.sort_values(by='match_score', ascending=False).head(top_n)

# --- Usage ---
recommendations = recommend_songs(user1, dtf2, scaler, xgb_vibe_model)
recommendations[['track_name', 'artists', 'match_score']]
```
